```python
operation = '+'
value_1 = 4
value_2 = 7

if operation == '+':
    print(value_1 + value_2)
elif operation == '-':
    print(value_1 - value_2)
elif operation == '*':
    print(value_1 * value_2)
elif operation == '/':
    print(value_1 / value_2)
```

```python
a = 1
b = 5
total = 0

if a > b:
    a, b = b, a

for i in range(a, b + 1):
    total += i

print(total)
```

```python
total = 0
count = 0

while True:
    grade = int(input())
    if grade == -1:
        break
    total += grade
    count += 1

avg = total / count

if avg >= 90:
    print("A")
elif avg >= 80:
    print("B")
elif avg >= 70:
    print("C")
elif avg >= 60:
    print("D")
else:
    print("F")
```
