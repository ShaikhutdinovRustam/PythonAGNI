```python
class Student:
    def __init__(self, name, group, grades, tel_number):
        self.name = name
        self.group = group
        self.grades = grades
        self.__tel_number = tel_number

    def add_grade(self, grade):
        self.grades.append(grade)

    def get_average(self):
        return sum(self.grades) / len(self.grades)

    def info(self):
        print(self.name, self.group, self.grades, self.__tel_number, self.get_average())


s1 = Student("Иван", "101", [5, 4], "89001112233")
s2 = Student("Анна", "102", [5, 5], "89004445566")
s3 = Student("Олег", "101", [3, 4], "89007778899")

s1.add_grade(5)
s2.add_grade(4)
s3.add_grade(5)

print(s1.get_average())
print(s2.get_average())
print(s3.get_average())

s1.info()
s2.info()
s3.info()
```
