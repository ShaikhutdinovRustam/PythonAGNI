```python
#№2
def calc(op, v_1, v_2):
    if op == '+':
        return v_1 + v_2
    elif op == '-':
        return v_1 - v_2
    elif op == '*':
        return v_1 * v_2
    elif op == '/':
        return v_1 / v_2
print(calc('+', 4, 7)) # 11
print(calc('-', 15, 18)) # -3
print(calc('*', 5, 5)) # 25
print(calc('/', 49, 7)) # 7.0
#№4
def quart(m):
    if 1<=m<=3:
        return "1 квартал"
    elif 4<=m<=6:
        return "2 квартал"
    elif 7<=m<=9:
        return "3 квартал"
    elif 10<=m<=12:
        return "4 квартал"
    else:
        return "ошибка"
def seas(m):
    if m in (12,1,2):
        return "зима"
    elif m in (3,4,5):
        return "весна"
    elif m in (6,7,8):
        return "лето"
    elif m in (9,10,11):
        return "осень"
    else:
        return "ошибка"
m=int(input())
s=seas(m)
q=quart(m)
print(m,q,s)
```

```python
def schet(text,bykvi):
    rez={}
    for b in bykvi:
        rez[b]=text.count(b)
    return rez
text="мама я дома"
bykvi=["м","а","я"]

itog=schet(text,bykvi)
for b,k in itog.items():
    print(b,k)
```
