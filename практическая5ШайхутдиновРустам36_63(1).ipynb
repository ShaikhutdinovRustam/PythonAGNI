```python
import json

try:
    with open('students.json', 'r', encoding='utf-8') as f:
        data = json.load(f)

    report_list = []
    all_sum = 0
    all_count = 0

    for item in data:
        if not isinstance(item, dict) or 'name' not in item or 'grades' not in item:
            raise KeyError("Отсутствуют ключи 'name' или 'grades'")

        name = item['name']
        grades = item['grades']

        if not isinstance(name, str) or not isinstance(grades, list):
            raise TypeError("Неверный тип данных в имени или оценках")

        for g in grades:
            if not isinstance(g, (int, float)):
                raise TypeError("Оценка должна быть числом")

        if len(grades) > 0:
            student_avg = sum(grades) / len(grades)
        else:
            student_avg = 0

        report_list.append({
            'name': name,
            'avg': round(student_avg, 2)
        })

        all_sum += sum(grades)
        all_count += len(grades)

    if all_count > 0:
        total_avg = all_sum / all_count
    else:
        total_avg = 0

    res = {
        'students': report_list,
        'total_avg': round(total_avg, 2)
    }

    with open('report.json', 'w', encoding='utf-8') as f:
        json.dump(res, f, ensure_ascii=False, indent=4)

    print("Файл report.json успешно создан")

except FileNotFoundError:
    print("Ошибка: Файл students.json не найден")
except KeyError as e:
    print(f"Ошибка: {e}")
except TypeError as e:
    print(f"Ошибка: {e}")
except json.JSONDecodeError:
    print("Ошибка: Некорректный формат JSON")
```
