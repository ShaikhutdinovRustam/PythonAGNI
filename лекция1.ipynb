```python
n = int(input("Сколько человек? "))
summa = 0

for i in range(n):
    vozrast = int(input("Возраст: "))
    summa = summa + vozrast

srednee = summa / n
print("Средний возраст:", srednee)
```

```python
color = input().lower()

r = int(color[0:2], 16)
g = int(color[2:4], 16)
b = int(color[4:6], 16)

if color == "000000":
    print("черный")
elif color == "ffffff":
    print("белый")
elif r == g == b:
    print("серый")
elif r >= g and r >= b:
    print("красный")
elif g >= r and g >= b:
    print("зеленый")
else:
    print("синий")
```
